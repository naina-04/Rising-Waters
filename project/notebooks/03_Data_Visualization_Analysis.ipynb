## 3. Data Visualizations

### Histogram
Histograms help us understand the distribution of a single numerical variable by grouping values into bins. Below, we plot the histogram for `Annual_Rainfall`.

```python
plt.figure(figsize=(10, 6))
sns.histplot(df['Annual_Rainfall'], bins=30, kde=False, color='skyblue')
plt.title('Histogram of Annual Rainfall')
plt.xlabel('Annual Rainfall')
plt.ylabel('Frequency')
plt.show()
```

**Explanation:** This histogram shows the frequency of different annual rainfall amounts. It helps us quickly identify the most common ranges of rainfall and spot any obvious skewness or anomalies in the data.

### Distribution Plot
A distribution plot provides a smoother, continuous view of the data distribution compared to a histogram, using Kernel Density Estimation (KDE).

```python
plt.figure(figsize=(10, 6))
sns.histplot(df['Annual_Rainfall'], bins=30, kde=True, color='salmon')
plt.title('Distribution of Annual Rainfall')
plt.xlabel('Annual Rainfall')
plt.ylabel('Density')
plt.show()
```

**Explanation:** The distribution plot adds a KDE curve, smoothing out the histogram bins to illustrate the probability density of the rainfall data. This gives a clearer picture of the overall shape of the distribution.

### Box Plot
Box plots are excellent for identifying outliers and understanding the spread (variance) of the data across different categories.

```python
plt.figure(figsize=(10, 6))
sns.boxplot(x='Flood', y='Annual_Rainfall', data=df, palette='Set2')
plt.title('Box Plot of Annual Rainfall by Flood Status')
plt.xlabel('Flood (0 = No, 1 = Yes)')
plt.ylabel('Annual Rainfall')
plt.show()
```

**Explanation:** This box plot compares the `Annual_Rainfall` for instances where floods occurred (1) vs did not occur (0). The box represents the interquartile range (middle 50% of data), the line inside is the median, and the whiskers show the range excluding outliers. It visually demonstrates whether higher rainfall strongly correlates with flooding.

### Correlation Matrix and Heatmap
A correlation matrix calculates the pairwise correlation coefficients between all numerical variables. A heatmap is a color-coded representation of this matrix.

```python
plt.figure(figsize=(20, 16))
corr_matrix = df.corr()
sns.heatmap(corr_matrix, annot=False, cmap='coolwarm', linewidths=0.5)
plt.title('Correlation Matrix Heatmap')
plt.show()
```

**Explanation:** The heatmap visualizes the correlation between all features. Colors closer to dark red indicate a strong positive correlation, while colors closer to dark blue indicate a strong negative correlation. By looking at the row/column for the `Flood` target, we can identify which features are the strongest predictors of flooding.

### Pairplot
A pairplot displays pairwise relationships across multiple variables in a grid format. Since plotting all 26 variables would be unreadable, we select a subset of key meteorological features.

```python
# Selecting a subset of features for the pairplot
subset_cols = ['Annual_Rainfall', 'Cloud_Visibility', 'Temperature', 'Monsoon_Intensity', 'Flood']
sns.pairplot(df[subset_cols], hue='Flood', palette='husl', diag_kind='kde')
plt.show()
```

**Explanation:** The pairplot creates scatterplots for each pair of selected variables, colored by the `Flood` outcome. The diagonal shows the density distribution for each variable separated by class. It is highly effective for spotting complex, non-linear relationships and seeing how well the features separate the flood vs. non-flood classes.
